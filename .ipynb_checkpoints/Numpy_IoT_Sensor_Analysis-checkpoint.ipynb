```python
import numpy as np
```

# Partie 1: Listes Python vs Tableaux NumPy

```python
# 1) Liste Python
temperatures_liste=[24.5, 25.1, 26.3, 27.0, 25.8]

# 2) Tableau NumPy équivalent
temperatures_np=np.array(temperatures_liste)

# 3) Multiplication par 2
resultat_liste=temperatures_liste * 2  # duplique la liste
resultat_np=temperatures_np * 2        # multiplie chaque valeur

print("Liste x2 : ", resultat_liste)
print("NumPy x2 : ", resultat_np)

  # 4) Différence observée : 
  #    - Sur une liste Python, `* 2` duplique la liste (concaténation). 
  #    - Sur un tableau NumPy, `* 2` multiplie chaque élément (opération vectorisée, élément par élément).
```

# Partie 2 : Création des données

```python
# 1) 10 mesures de température initialisées à zéro
temp_zeros=np.zeros(10)

# 2) 10 valeurs égales à 1
temp_ones=np.ones(10)

# 3) 10 valeurs initialisées à 25
temp_25=np.full(10, 25)

# 4) Matrice identité 5x5
identite=np.eye(5)

# 5) Les 24 heures d'une journée (0 à 23)
heures=np.arange(24)

# 6) 10 valeurs régulièrement réparties entre 20°C et 30°C
temp_reparties=np.linspace(20, 30, 10)

# 7) 100 températures aléatoires reproductibles
rng=np.random.default_rng(seed=42)
temp_aleatoires=rng.normal(loc=25, scale=3, size=100).round(1)

print("Zéros : ", temp_zeros)
print("Uns : ", temp_ones)
print("25 :", temp_25)
print("Identité :\n", identite)
print("Heures :", heures)
print("Réparties 20-30 :", temp_reparties)
print("Aléatoires (extrait) :", temp_aleatoires[:10])
```

# Partie 3 : Exploration des tableaux

```python
# 1) 24 températures aléatoires d'une journée
rng=np.random.default_rng(seed=1)
temperatures_jour=rng.normal(loc=25, scale=3, size=24).round(1)

print("Températures: ", temperatures_jour)

# 2) Propriétés du tableau
print("a) Nombre de dimensions :", temperatures_jour.ndim)
print("b) Nombre de valeurs :", temperatures_jour.size)
print("c) Type des données :", temperatures_jour.dtype)
print("d) Forme du tableau :", temperatures_jour.shape)
```

# Partie 4 : Indexation et slicing

```python
# 1) Première mesure
print("Première mesure :", temperatures_jour[0])

# 2) Dernière mesure
print("Dernière température :", temperatures_jour[-1])

# 3) Température de 8h
print("Température à 8h :", temperatures_jour[8])

# 4) Températures de 8h à 12h
print("Températures 8h-12h :", temperatures_jour[8:13])

# 5) Températures des 6 premières heures
print("6 premières heures :", temperatures_jour[:6])

# 6) Températures des heures paires
print("Heures paires :", temperatures_jour[::2])
```

# Partie 5 : Filtrage booléen

```python
# 1) Températures supérieures à 30°C
print("Températures > 30°C :", temperatures_jour[temperatures_jour > 30])

# 2) Températures "normales" (entre 20°C et 30°C)
normales = temperatures_jour[(temperatures_jour >= 20) & (temperatures_jour <= 30)]
print("Températures normales (20-30°C) :", normales)

# 3) Tableau booléen des anomalies (> 35°C)
anomalies = temperatures_jour > 35
print("Anomalies (booléen) :", anomalies)

# 4) Nombre de températures anormales
print("Nombre d'anomalies :", anomalies.sum())
```

# Partie 6 : Manipulation des dimensions (5 capteurs)

```python
# 1) Matrice des températures des 5 capteurs sur 24h
rng = np.random.default_rng(seed=2)
temp_5_capteurs = rng.normal(loc=25, scale=3, size=(5, 24)).round(1)
print("Matrice (5 capteurs x 24h) :\n", temp_5_capteurs)
print("Shape :", temp_5_capteurs.shape)

# 2) Transformation en tableau unidimensionnel
temp_flatten = temp_5_capteurs.flatten()
print("Aplati, shape :", temp_flatten.shape)

# 3) Transposée de la matrice
temp_transposee = temp_5_capteurs.T
print("Transposée, shape :", temp_transposee.shape)

# 4) Toutes les mesures du capteur 1
print("Capteur 1 :", temp_5_capteurs[0, :])

# 5) Toutes les mesures du capteur 5
print("Capteur 5 :", temp_5_capteurs[4, :])

# 6) Mesures de tous les capteurs à 10h
print("Tous les capteurs à 10h :", temp_5_capteurs[:, 10])

# 7) Mesures des capteurs 2 à 4 entre 8h et 12h
print("Capteurs 2-4, 8h-12h :\n", temp_5_capteurs[1:4, 8:13])
```

# Partie 7 : Concaténation (2e bâtiment, 3 capteurs)

```python
# 1) Matrice des températures des 3 capteurs du 2e bâtiment
rng = np.random.default_rng(seed=3)
temp_3_capteurs = rng.normal(loc=25, scale=3, size=(3, 24)).round(1)
print("Bâtiment 2 (3 capteurs x 24h) :\n", temp_3_capteurs)

# 2) Fusion des 8 capteurs (5 + 3), deux méthodes équivalentes
tous_capteurs_v1 = np.concatenate([temp_5_capteurs, temp_3_capteurs], axis=0)  # Méthode 1
tous_capteurs_v2 = np.vstack([temp_5_capteurs, temp_3_capteurs])               # Méthode 2

print("Dimensions après fusion (concatenate) :", tous_capteurs_v1.shape)
print("Dimensions après empilement (vstack) :", tous_capteurs_v2.shape)
print("Résultats identiques :", np.array_equal(tous_capteurs_v1, tous_capteurs_v2))
```

# Partie 8 : Calcul scientifique (24 températures du bâtiment 1)

```python
# 1) Statistiques descriptives
print("Min :", temperatures_jour.min())
print("Max :", temperatures_jour.max())
print("Somme :", temperatures_jour.sum())
print("Moyenne :", temperatures_jour.mean().round(2))
print("Médiane :", np.median(temperatures_jour))
print("Variance :", temperatures_jour.var().round(2))
print("Écart-type :", temperatures_jour.std().round(2))
```

```python
# 2) Standardisation : (x - moyenne) / écart-type
temp_standardisees = (temperatures_jour - temperatures_jour.mean()) / temperatures_jour.std()
print("Standardisées (extrait) :", temp_standardisees.round(2)[:6])
print("Moyenne après standardisation :", temp_standardisees.mean().round(6))
print("Écart-type après standardisation :", temp_standardisees.std().round(6))
```

```python
# 3) Normalisation Min-Max : (x - min) / (max - min)
temp_normalisees = (temperatures_jour - temperatures_jour.min()) / (temperatures_jour.max() - temperatures_jour.min())
print("Normalisées (extrait) :", temp_normalisees.round(2)[:6])
print("Min après normalisation :", temp_normalisees.min())
print("Max après normalisation :", temp_normalisees.max())
```

```python
# 4) Conversion en Fahrenheit : F = C * 9/5 + 32
temp_fahrenheit = temperatures_jour * 9/5 + 32
print("Températures en Fahrenheit :", temp_fahrenheit.round(1))
```

# Partie 9 : Broadcasting

```python
# 1) Matrice des températures à 10h, 11h, 12h dans le 2e bâtiment (3 capteurs)
rng = np.random.default_rng(seed=4)
temperatures_3x3 = rng.normal(loc=25, scale=2, size=(3, 3)).round(1)
# lignes = 10h, 11h, 12h | colonnes = Capteur 1, Capteur 2, Capteur 3
print("Températures 10h/11h/12h (3 heures x 3 capteurs) :\n", temperatures_3x3)
```

```python
# 2) Correction de calibration par capteur, appliquée automatiquement (broadcasting)
correction = np.array([1, -0.5, 0.2])
mesures_corrigees = temperatures_3x3 + correction

print("Correction par capteur :", correction)
print("Mesures corrigées :\n", mesures_corrigees)
```
