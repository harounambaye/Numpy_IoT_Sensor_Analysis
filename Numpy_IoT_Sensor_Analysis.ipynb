```python
import numpy as np
```

# Partie 1: Listes Python vs Tableaux NumPy

```python
# 1) Liste Python
temperatures_liste=[24.5, 25.1, 26.3, 27.0, 25.8]

# 2) Tableau NumPy équivalent
temperatures_np=np.array(temperatures_liste)

# 3) Multiplication par 2
resultat_liste=temperatures_liste * 2  # duplique la liste
resultat_np=temperatures_np * 2        # multiplie chaque valeur

print("Liste x2 : ", resultat_liste)
print("NumPy x2 : ", resultat_np)

  # 4) Différence observée : 
  #    - Sur une liste Python, `* 2` duplique la liste (concaténation). 
  #    - Sur un tableau NumPy, `* 2` multiplie chaque élément (opération vectorisée, élément par élément).''
```

### 4) Différence observée : sur une liste Python, `* 2` **duplique** la liste (concaténation). Sur un tableau NumPy, `* 2` **multiplie chaque élément** (opération vectorisée, élément par élément).
